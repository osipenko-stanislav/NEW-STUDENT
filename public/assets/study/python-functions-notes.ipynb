# Конспект: функции в Python

Примеры функций, аргументов и обработки строк.

Реализуйте функцию, возвращающую длины слов, и проверьте пограничные случаи.

```python
def word_lengths(text):
    return [len(word) for word in text.split()]

assert word_lengths("hello world") == [5, 5]
assert word_lengths("") == []
```
